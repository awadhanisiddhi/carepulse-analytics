```python
financial_summary = (
    admissions_df[
        ["admission_id", "department_id"]
    ]
    .merge(
        billing_df[
            ["admission_id", "total_bill", "insurance_covered"]
        ],
        on="admission_id",
        how="left"
    )
    .merge(
        departments_df[
            ["department_id", "department_name"]
        ],
        on="department_id",
        how="left"
    )
    .groupby(
        ["department_id", "department_name"]
    )
    .agg(
        total_revenue=("total_bill", "sum"),
        total_insurance_covered=("insurance_covered", "sum"),
        average_bill=("total_bill", "mean")
    )
    .sort_values("total_revenue", ascending=False)
)

financial_summary
```

```python
department_admissions = (
    admissions_df
    .groupby("department_id")
    .agg(
        total_admissions=("admission_id", "count"),
        average_waiting_time=("waiting_time_minutes", "mean"),
        average_length_of_stay=("length_of_stay", "mean")
    )
    .sort_values("total_admissions", ascending=False)
)

department_admissions
```
